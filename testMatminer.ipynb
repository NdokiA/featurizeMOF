```python
#Python 3.11
from workFunc.matminerFeaturize import featurizer
import os
```

```python
os.makedirs("mofs", exist_ok=True)
#taruh file cif MOFs di sini
```

```python
#Test MOFs
fz = featurizer(cif_dir="mofs", max_sites=3000)
target = fz.discover_target()

feature_df = fz.featurize(target["name"])
merged_df = fz.attach(target, feature_df)
numeric_df = fz.numeric_matrix(merged_df)

print(merged_df.shape, "->", numeric_df.shape)
```

```python
#Save dataset
os.makedirs("Dataset", exist_ok = True)
merged_df.to_csv("Dataset/processedCIF.csv") 
numeric_df.to_csv("Dataset/numericCIF.csv")
```
